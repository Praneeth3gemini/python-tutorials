```python
# Python Strings: Practice Workbook

# 1. Create a string
text = "hello PYTHON world"
print("1. Original:", text)

# 2. Indexing
print("2. First character:", text[0])
print("2. Last character:", text[-1])

# 3. Slicing
print("3. First 5 characters:", text[:5])
print("3. Reversed:", text[::-1])

# 4. String methods
print("4. Upper:", text.upper())
print("4. Lower:", text.lower())
print("4. Title:", text.title())
print("4. Replace:", text.replace("world", "students"))

# 5. Split and join
words = text.split()
print("5. Split words:", words)
print("5. Joined with hyphen:", "-".join(words))

# 6. Capitalize each word without using title()
capitalized_words = []
for word in words:
    capitalized_words.append(word[0].upper() + word[1:].lower())
print("6. Manually Capitalized:", " ".join(capitalized_words))
```

# Detailed Explanation of Python Strings

In Python, a **string** is a sequence of Unicode characters. Strings are **immutable**, meaning that once a string is created, its characters cannot be modified or replaced in place. Any string manipulation method will instead return a new string.

### Character Encoding & Representation
Python strings use Unicode (UTF-8 by default in Python 3), meaning they can store characters from any language, along with emojis, mathematical symbols, and special characters.

### String Formatting & Escape Sequences
- **Literal formatting (f-strings)** allows direct interpolation of expressions inside `{}`.
- **Escape characters** (like `\n` for newline, `\t` for tab) let you embed special formatting inside standard text literals.

---

## Summary of Python String Methods

| Method | Description | Return Type | Example |
| :--- | :--- | :--- | :--- |
| `.upper()` | Converts all characters to uppercase | `str` | `"py".upper() -> "PY"` |
| `.lower()` | Converts all characters to lowercase | `str` | `"PY".lower() -> "py"` |
| `.title()` | Capitalizes first letter of every word | `str` | `"hi there".title() -> "Hi There"` |
| `.capitalize()`| Capitalizes only the first character | `str` | `"hello".capitalize() -> "Hello"` |
| `.strip()` | Removes leading and trailing whitespace | `str` | `"  ok  ".strip() -> "ok"` |
| `.lstrip()` | Removes leading whitespace | `str` | `"  ok  ".lstrip() -> "ok  "` |
| `.rstrip()` | Removes trailing whitespace | `str` | `"  ok  ".rstrip() -> "  ok"` |
| `.find(sub)` | Returns index of substring or `-1` if not found | `int` | `"abc".find("b") -> 1` |
| `.count(sub)`| Counts occurrences of substring | `int` | `"banana".count("a") -> 3` |
| `.split(sep)`| Splits string into a list of substrings | `list` | `"a b".split() -> ["a", "b"]` |
| `.join(seq)` | Joins an iterable of strings with separator | `str` | `"-".join(["a", "b"]) -> "a-b"` |
| `.replace(o, n)`| Replaces old substring with new substring | `str` | `"abc".replace("b", "z") -> "azc"` |
| `.isdigit()` | Returns `True` if string contains only digits | `bool` | `"123".isdigit() -> True` |
| `.isalpha()` | Returns `True` if string contains only alphabets | `bool` | `"abc".isalpha() -> True` |

```python
# 1. Convert a string to uppercase.
print("1:", "python".upper())

# 2. Convert a string to lowercase.
print("2:", "HELLO".lower())

# 3. Remove spaces from both ends.
print("3:", " hello ".strip())

# 4. Count the characters in a string.
print("4:", len("python"))

# 5. Print the first and last characters.
s5 = "python"
print("5:", s5[0], s5[-1])

# 6. Reverse a string using slicing.
print("6:", "hello"[::-1])

# 7. Check whether a substring exists.
print("7:", "tho" in "python")

# 8. Replace every space with a hyphen.
print("8:", "learn python".replace(" ", "-"))

# 9. Count how many times a character appears.
print("9:", "banana".count("a"))

# 10. Extract the first three characters.
print("10:", "programming"[:3])
```

```python
# 11. Capitalize the first letter.
print("11:", "python".capitalize())

# 12. Convert each word to title case.
print("12:", "learn python".title())

# 13. Count the number of words.
print("13:", len("learn python today".split()))

# 14. Join words using an underscore.
print("14:", "_".join("learn python".split()))

# 15. Remove all spaces from a string.
print("15:", "p y t h o n".replace(" ", ""))

# 16. Reverse each word but keep word order.
print("16:", " ".join([w[::-1] for w in "hello world".split()]))

# 17. Count vowels (a, e, i, o, u).
s17 = "education"
print("17:", sum(1 for char in s17 if char.lower() in "aeiou"))

# 18. Count consonants in an English-letter string.
s18 = "python"
print("18:", sum(1 for char in s18 if char.isalpha() and char.lower() not in "aeiou"))

# 19. Count uppercase and lowercase letters separately.
s19 = "PyThOn"
uppers = sum(1 for c in s19 if c.isupper())
lowers = sum(1 for c in s19 if c.islower())
print("19:", f"{uppers} uppercase, {lowers} lowercase")

# 20. Remove duplicate characters, preserving first occurrences.
s20 = "programming"
seen = []
for char in s20:
    if char not in seen:
        seen.append(char)
print("20:", "".join(seen))
```

```python
# 21. Check whether a string is a palindrome.
s21 = "madam"
print("21:", s21 == s21[::-1])

# 22. Swap uppercase to lowercase and lowercase to uppercase.
print("22:", "PyThOn".swapcase())

# 23. Find the first occurrence of a character without using find().
s23 = "banana"
target = "n"
idx = -1
for i, char in enumerate(s23):
    if char == target:
        idx = i
        break
print("23:", idx)

# 24. Count occurrences of a substring without using count().
s24 = "banana"
sub = "an"
cnt = 0
for i in range(len(s24) - len(sub) + 1):
    if s24[i:i+len(sub)] == sub:
        cnt += 1
print("24:", cnt)

# 25. Remove every occurrence of a chosen character.
print("25:", "banana".replace("a", ""))

# 26. Keep only the digits in a string.
print("26:", "".join([c for c in "a1b2c3" if c.isdigit()]))

# 27. Keep only the letters in a string.
print("27:", "".join([c for c in "py3th0on!" if c.isalpha()]))

# 28. Check whether two strings are equal ignoring case.
print("28:", "Python".lower() == "python".lower())

# 29. Find the longest word in a sentence.
s29 = "I love programming"
print("29:", max(s29.split(), key=len))

# 30. Make the first and last character uppercase.
s30 = "python"
print("30:", s30[0].upper() + s30[1:-1] + s30[-1].upper())
```

```python
# 31. Find the most frequent character without using a dictionary.
s31 = "banana"
most_freq = max(s31, key=s31.count)
print("31:", most_freq)

# 32. Remove consecutive duplicate characters.
s32 = "aaabbcdaa"
res32 = []
for char in s32:
    if not res32 or res32[-1] != char:
        res32.append(char)
print("32:", "".join(res32))

# 33. Compress repeated consecutive characters.
s33 = "aaabbc"
compressed = []
count = 1
for i in range(1, len(s33)):
    if s33[i] == s33[i-1]:
        count += 1
    else:
        compressed.append(f"{s33[i-1]}{count}")
        count = 1
compressed.append(f"{s33[-1]}{count}")
print("33:", "".join(compressed))

# 34. Check whether one string is a rotation of another.
s34_1, s34_2 = "abcd", "cdab"
print("34:", len(s34_1) == len(s34_2) and s34_2 in (s34_1 + s34_1))

# 35. Find all characters that occur in both strings, without duplicates.
s35_1, s35_2 = "apple", "plane"
both = []
for char in s35_1:
    if char in s35_2 and char not in both:
        both.append(char)
print("35:", "".join(both))

# 36. Find the longest common prefix of two strings.
s36_1, s36_2 = "flower", "flow"
prefix = ""
for i in range(min(len(s36_1), len(s36_2))):
    if s36_1[i] == s36_2[i]:
        prefix += s36_1[i]
    else:
        break
print("36:", prefix)

# 37. Remove punctuation using a loop and a chosen punctuation string.
s37 = "hi, there!"
punc = ",.!?:;"
print("37:", "".join([c for c in s37 if c not in punc]))

# 38. Check whether a sentence contains every vowel.
s38 = "Education is a unique resource"
vowels_present = all(v in s38.lower() for v in "aeiou")
print("38:", vowels_present)

# 39. Reverse the order of words in a sentence.
print("39:", " ".join("learn python today".split()[::-1]))

# 40. Find the second-longest word in a sentence.
s40 = "I love learning python"
words_sorted = sorted(list(set(s40.split())), key=len, reverse=True)
print("40:", words_sorted[1] if len(words_sorted) > 1 else "")
```

```python
# 41. Make an alternating uppercase/lowercase string.
s41 = "python"
alternating = "".join([char.upper() if i % 2 == 0 else char.lower() for i, char in enumerate(s41)])
print("41:", alternating)

# 42. Remove duplicate words while preserving their first occurrence.
s42 = "go go learn python learn"
seen_words = []
for w in s42.split():
    if w not in seen_words:
        seen_words.append(w)
print("42:", " ".join(seen_words))

# 43. Check whether two strings are anagrams, ignoring case and spaces.
s43_1, s43_2 = "Listen", "Silent"
clean1 = sorted(s43_1.replace(" ", "").lower())
clean2 = sorted(s43_2.replace(" ", "").lower())
print("43:", clean1 == clean2)

# 44. Find the longest run of the same consecutive character.
s44 = "aabbbbcc"
max_run = ""
current_run = ""
for char in s44:
    if not current_run or char == current_run[-1]:
        current_run += char
    else:
        if len(current_run) > len(max_run):
            max_run = current_run
        current_run = char
if len(current_run) > len(max_run):
    max_run = current_run
print("44:", max_run)

# 45. Replace multiple spaces with a single space without using split().
s45 = "learn     python"
while "  " in s45:
    s45 = s45.replace("  ", " ")
print("45:", s45)

# 46. Create an acronym from a sentence.
s46 = "self contained underwater breathing apparatus"
print("46:", "".join([w[0].upper() for w in s46.split()]))

# 47. Check if a string contains only digits without using isdigit().
s47 = "12345"
only_digits = all(char in "0123456789" for char in s47)
print("47:", only_digits)

# 48. Find the character at the middle index; handle even lengths.
s48 = "python"
mid = len(s48) // 2
if len(s48) % 2 == 0:
    print("48:", s48[mid-1:mid+1])
else:
    print("48:", s48[mid])

# 49. Convert a sentence to title case without using title().
s49 = "learn python well"
print("49:", " ".join([w[0].upper() + w[1:].lower() if len(w) > 0 else "" for w in s49.split()]))
```
